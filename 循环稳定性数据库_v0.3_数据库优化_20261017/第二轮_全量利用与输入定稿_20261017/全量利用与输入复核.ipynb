从本目录打开。SQL均mode=ro，输出为本轮实际执行结果；不含研究训练。

```python
from pathlib import Path
import sqlite3,json
r=Path.cwd()
c=sqlite3.connect((r/'cycling_v0_3r2.sqlite').as_uri()+'?mode=ro',uri=True)
c.row_factory=sqlite3.Row
print(c.execute('pragma integrity_check').fetchone()[0])
```

```python
print(json.dumps([dict(x) for x in c.execute('select track,count(*) endpoints from utilization_routes group by track')],ensure_ascii=False,indent=2))
```

```python
print(json.dumps([dict(x) for x in c.execute("select record_id,mass_basis_checked,eligible from admissions where mass_basis_checked=1")],ensure_ascii=False,indent=2))
print('approved',c.execute('select count(*) from training_ready').fetchone()[0])
```

```python
print(json.dumps([dict(x) for x in c.execute('select candidate_id,paper_id,system,metric_kind,cycle_count,retention_pct,eligible from staged_recovery_endpoints')],ensure_ascii=False,indent=2))
c.close()
```
