# 数据库复核
此笔记本仅查询当前版本数据库；已保存的输出来自本次实际SQLite查询。无训练。

```python
import pathlib,sqlite3,json
ROOT=pathlib.Path('D:\\机器学习-超级电容器\\循环稳定性数据库_v0.1_20261017')
c=sqlite3.connect((ROOT/'cycling_v0_1.sqlite').as_uri()+'?mode=ro',uri=True)
c.row_factory=sqlite3.Row
# 层级计数
sql="SELECT 'papers' name,count(*) n FROM papers UNION ALL SELECT 'materials',count(*) FROM materials UNION ALL SELECT 'cycling_tests',count(*) FROM cycling_tests UNION ALL SELECT 'endpoints',count(*) FROM cycling_endpoints UNION ALL SELECT 'training_ready',count(*) FROM training_ready"
print(json.dumps([dict(r) for r in c.execute(sql)],ensure_ascii=False,indent=2))
```

```python
# 主体系队列
sql='SELECT paper_id,count(*) endpoints,count(distinct material_id) materials FROM primary_screening_candidates GROUP BY paper_id ORDER BY paper_id'
print(json.dumps([dict(r) for r in c.execute(sql)],ensure_ascii=False,indent=2))
```

```python
# 原值和标签类型
sql='SELECT retention_relation,count(*) n,sum(retention_pct>100) above100 FROM cycling_endpoints GROUP BY retention_relation'
print(json.dumps([dict(r) for r in c.execute(sql)],ensure_ascii=False,indent=2))
```
