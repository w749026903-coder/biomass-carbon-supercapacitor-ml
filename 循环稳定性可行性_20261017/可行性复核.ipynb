# 循环稳定性数据可行性复核
本地候选清单的计数与分组检查。结果不是训练样本准入。

```python
from pathlib import Path
import json, collections
base = Path.cwd()
if not (base/'循环稳定性逐记录候选.json').exists():
    base = Path('D:/机器学习-超级电容器/循环稳定性可行性_20261017')
rows=json.loads((base/'循环稳定性逐记录候选.json').read_text(encoding='utf-8'))
def count(items):
    return {'终点':len(items),'暂定材料':len({r['material_id'] for r in items}),'论文':len({r['paper_id'] for r in items})}
gcd=[r for r in rows if r['candidate_clean'] and r['method']=='GCD' and r['system'] in ['3E','2E_symmetric']]
print('全部候选',count(rows))
print('明确GCD碳材料候选',count(gcd))
print('基本条件配齐',count([r for r in rows if r['provisional_core_conditions']]))
print('分体系', {s:count([r for r in gcd if r['system']==s]) for s in ['3E','2E_symmetric']})
print('正式准入',sum(r['eligible_for_database'] for r in rows))
assert len(rows)==81
assert not any(r['eligible_for_database'] for r in rows)
assert len({r['material_id'] for r in rows if r['paper_id']=='R107'})==1
```
