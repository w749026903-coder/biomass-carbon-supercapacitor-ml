```python
import pathlib, sqlite3
# Open the notebook from its containing v0.2 directory.
db = pathlib.Path('cycling_v0_2.sqlite').resolve()
conn = sqlite3.connect(db.as_uri()+'?mode=ro', uri=True)
conn.row_factory = sqlite3.Row
```

```python
sql = "SELECT 'candidate_endpoints' AS kind,count(*) AS n FROM endpoint_review UNION ALL SELECT 'training_ready',count(*) FROM training_ready"
rows = [dict(x) for x in conn.execute(sql)]
print(rows)
```

```python
sql = 'SELECT status,count(*) AS endpoints FROM admissions WHERE source_checked=1 GROUP BY status'
rows = [dict(x) for x in conn.execute(sql)]
print(rows)
```

```python
sql = "SELECT p.paper_id,r.record_id,r.field_name,r.decision,r.locator FROM field_review_latest r JOIN raw_records p ON p.record_id=r.record_id WHERE r.decision IN ('conflict_text_vs_cycle_inset','requires_missing_SI','material_scope_review') ORDER BY r.record_id,r.field_name"
rows = [dict(x) for x in conn.execute(sql)]
print(rows)
```

```python
sql = 'SELECT relationship_id,paper_id,related_doi,relationship,status,note FROM source_relationships ORDER BY relationship_id'
rows = [dict(x) for x in conn.execute(sql)]
print(rows)
```
